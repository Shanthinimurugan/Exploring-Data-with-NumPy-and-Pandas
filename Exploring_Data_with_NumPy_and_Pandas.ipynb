### Numpy Array Operations:

```python
!pip install numpy
```

```python
import numpy as np
```

```python
# 1. Create a 1D NumPy Array:

temperatures_w1 = np. array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

```python
# 2. Inspection and Properties:
# ● Inspect the shape, data type, and the number of elements in the array

print("Shape: ",temperatures_w1.shape)
print("Object data type: ",type(temperatures_w1))
print("Element data type: ", temperatures_w1.dtype)
print("Dimension: ",temperatures_w1.ndim)
print("Number of Elements: ",temperatures_w1.size)
```

```python
# 3. Array Operations:
#  Convert the temperatures_w1 from Celsius to Fahrenheit
# Fahrenheit = (Celsius * 9/5) + 32

Fahrenheit = (temperatures_w1*9/5)+32
print("Fahrenheit: ",Fahrenheit)

# ● Find the maximum, minimum, and mean temperatures for the week.

print("Mean temperatures: ",round(np.mean(temperatures_w1), 2))
print("Minimum temperature: ",np.min(temperatures_w1))
print("Maximum temperature: ",np.max(temperatures_w1))
```

```python
# 4. Array Slicing and Indexing:
# ● Extract the temperatures for the first three days of the week.

print("First three days of teperatures: ",temperatures_w1[0:3].tolist())
# [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

# ● Extract the temperatures for the weekend (last two days).

print("Weekend temperatures: ",temperatures_w1[-2:].tolist())

# ● Extract the temperatures for the middle three days of the week.

print("Middle three days of temperatures: ",temperatures_w1[2:5].tolist())
```

```python
# 5. Create a 2D Array:
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                        [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
temperatures
```

```python
# 6. Inspect and Slice the 2D Array:
# ● Inspect and print the shape, data type, and total number of elements in the temperatures array.

print("Shape: ",temperatures.shape)
print("Object data type: ",type(temperatures))
print("Element data type: ",temperatures.dtype)
print("Dimension: ",temperatures.ndim)
print("Number of Elements: ",temperatures.size)
```

```python
# ● Extract the temperatures for each week and the weekends (last two days) for both weeks
print("Week 1 temperatures: ",temperatures[0].tolist())
print("Week 2 temperatures: ",temperatures[1].tolist())
print("Weekend temperatures: ",temperatures[0:2,-2:].tolist())
```

### Pandas Series:

```python
!pip install pandas
```

```python
import pandas as pd
```

```python

# 1. Creating Pandas Series:

marks_ser = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'],name="marks")
marks_ser
```

```python
# 2. Indexing and Slicing:

# ● Use the integer index position to access the mark of the 1st rank student.

print("Mark of the 1st rank student: ",marks_ser.iloc[0],"\n")

# ● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.

print("Marks of the top 3 ranks:")
print(marks_ser.loc["Rank1":"Rank3"],"\n")

# ● Use the iloc accessor to retrieve the mark of the 3rd-ranked student.

print("Mark of the 3rd-ranked student: ",marks_ser.iloc[2],"\n")

# ● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
print("Marks greater than 90:")
print(marks_ser.loc[marks_ser>90])

```

```python
# 3. Manipulating Series:
# ● Modify the mark of the student with the 1st rank to 100.

marks_ser.loc["Rank1"]=100
marks_ser
```

```python
# ● Remove the entry corresponding to the last rank from the Series.
marks_ser.drop('Rank5',inplace=True)
marks_ser

```

```python
# ● Compute the CGPA by dividing each mark in the Series by 10.

CGPA = marks_ser/10
CGPA
```

### Pandas DataFrame:

```python
# 1. Creating Pandas DataFrame:

transactions = pd.DataFrame({
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    "Region": ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South',
'North'],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

```python
# 2. Data Exploration:
# ● Display the transactions DataFrame and its basic information, including head, tail,
# shape, column names, and data types.

transactions.head()

```

```python
transactions.tail()
```

```python
transactions.info()
```

```python
transactions.describe()
```

```python
print("Shape: ",transactions.shape)
print("Column Names: ",transactions.columns,"\n")
print(transactions.dtypes)
```

```python
# ● Display only the 'ProductCategory' and 'Amount' columns.

transactions[['ProductCategory', 'Amount']]
```

```python
# ● Retrieve the last 3 columns of the DataFrame using iloc or loc.

transactions.iloc[:,-3:]
# or
transactions.loc[:,"ProductCategory":"Amount"]
```

```python
# ● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.

transactions[(transactions['Region']=="North") & (transactions['Amount'] > 200)]
```

```python
# ● Find the value counts for the 'ProductCategory' column.

transactions["ProductCategory"].value_counts()
```

```python

# ● Find the unique values in the 'Region' column.

transactions["Region"].unique()
```

```python
# ● Group by 'Region' and find the mean amount for each region.

transactions.groupby("Region")["Amount"].mean()
```

```python
# 3. Manipulating the DataFrame:
# ● Modify the 'Amount' for the transaction with TransactionID 102 to 165.

transactions.loc[1,"Amount"]=165

# or
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165
transactions
```

```python
# ● Add a new column 'Discount' by calculating 10% of the 'Amount'.

transactions["Discount"]=(transactions["Amount"]*10)/100
transactions
```

```python
# ● Remove the row with TransactionID 109.

transactions.drop(transactions[transactions["TransactionID"] == 109].index, inplace= True)
transactions
```

```python
# ● Delete the 'Discount' column from the DataFrame.

transactions.drop("Discount",axis=1,inplace=True)
transactions
```
